#### Array creation

```python
import numpy as np
a1D = np.array([1, 2, 3, 4])
print('1D array:')
print(a1D)
a2D = np.array([[1, 2], [3, 4]])
print('2D array:')
print(a2D)
a3D = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print('3D array:')
print(a3D)
```

#### Creating arrays using arange

```python
# numpy.arange creates arrays with regularly incrementing values

import numpy as np
ar = np.arange(10)
print('1D array using arange:')
print(ar)
ar1 = np.arange(2, 10, dtype=int)
print('1D array with start, stop, and dtype:')
print(ar1)
ar3 = np.arange(2, 3, 0.1)
print('1D array with float step:')
print(ar3)
```

##### Common Array Initializations


```python
ar1 = np.zeros((2, 3))        # 2x3 array of zeros
print(ar1)
ar2 = np.ones((3, 3))         # 3x3 array of ones
print(ar2)

ar3 = np.arange(0, 10, 2)  # 1D array with step of 2
print(ar3)
```

##### Array Attributes

```python
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)      # (2, 3)
print(arr.ndim)       # 2
print(arr.size)       # 6
print(arr.dtype)      # data type (e.g., int32)
```

#### Array Indexing & Slicing

```python
# numpy indexing and slicing
a = np.array([10, 20, 30, 40, 50])
a[0]          # 10
a[1:4]        # [20, 30, 40]
a[-1]         # 50
```

####  Operations on Arrays

```python
# Element-wise operations:

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
c = a + b  # addition
print(c) # Output: [5 7 9]
d = a * b  # multiplication
print(d) # Output: [4 10 18]

# Matrix multiplication:
sum = np.dot(a, b)  # Output: (1*4 + 2*5 + 3*6) = 32
print(sum)  
```

#### Aggregate functions:

```python
a = np.array([1, 2, 3])
print(a.sum())
print(a.mean())
print(a.max())
print(a.min())
```

#### Reshaping

#### Reshaping

```python
# reshape() :  changes the shape of an array without changing its data

a = np.arange(6)     # [0, 1, 2, 3, 4, 5]
b = a.reshape((2, 3))
print('Reshaped array:')
print(b)  # Output: [[0, 1, 2], [3, 4, 5]]
```

#### Broadcasting

#### Scalar Broadcasting: A single value can be applied across an entire array.


```python
import numpy as np
arr = np.array([1, 2, 3])
print(arr + 5)  # Output: [6 7 8]


a1 = np.array([2, 4, 6])
a2 = np.array([[1, 3, 5], [7, 9, 11]])
print(a1 + a2)
```
