```python
# Given values
TP = 10
TN = 12
FP = 1
FN = 2

# Calculate metrics
accuracy = (TP + TN) / (TP + TN + FP + FN)
precision = TP / (TP + FP)
recall = TP / (TP + FN)
f1_score = 2 * (precision * recall) / (precision + recall)
fpr = FP / (FP + TN)

# Print results
print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1 Score:", f1_score)
print("False Positive Rate (FPR):", fpr)
```
