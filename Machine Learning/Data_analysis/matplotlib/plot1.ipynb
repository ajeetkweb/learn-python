```python
# import matplotlib library
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import style

web_customers = [123, 400, 600, 800, 1200, 500, 600, 1000, 900, 900]
time_hrs =      [8,    9,  10,  11,   12,  13,  14,  16,   17,  18]
```

```python
# select style of plot

style.use('ggplot')
plt.plot(time_hrs, web_customers)
plt.title('Web site Traffic')
plt.xlabel('Hours')
plt.ylabel('No of Customers')
plt.show()
```

```python
# initializing the data
x = [10, 20, 30, 40]
y = [20, 30, 40, 50]
  
# plotting the data
plt.plot(x, y,)
  
# Adding the title
plt.title("Simple Plot")
  
# Adding the labels
plt.ylabel("y-axis")
plt.xlabel("x-axis")
plt.show()
```

```python

x = np.linspace(0,5, 11)
x
```

```python
y = x **2
y
```

```python
plt.plot(x,y, 'b')
plt.xlabel('X Label')
plt.ylabel('Y Label')
plt.title("Line Plot")
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x, y, 'r')

plt.subplot(1,2,2)
plt.plot(y,x, 'b')

plt.show()
```

```python
# object oriented method to create plot


fig = plt.figure()

axes = fig.add_axes([.5,.5,1,1])
axes.plot(x,y)
axes.set_xlabel('X Label')
axes.set_ylabel('Y Label')
axes.set_title("Line plot")
```

```python
fig = plt.figure()

axes1 = fig.add_axes([0.1, 0.1, 0.8, 0.8])
axes2 = fig.add_axes([0.2, 0.4, 0.3, 0.3])

axes1.plot(x,y)

axes2.plot(y,x)
```
