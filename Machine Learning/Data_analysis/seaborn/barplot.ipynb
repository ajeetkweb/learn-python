## sns.barplot()

### Syntax : sns.barplot(x=None, y=None, hue=None, data=None)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../Titanic-Dataset.xls")
df
```

```python
sns.barplot(x='Pclass', y='Fare', data = df)
plt.show()
```

```python
sns.barplot(x= 'Pclass', y= 'Fare', hue = 'Sex', data = df)
plt.show()
```

```python
sns.barplot(x= 'Pclass', y = 'Survived', data = df)
plt.show()
```

```python
# Initialize data
State = ["Rajasthan", "Bihar", "Madhya Pradesh",
		"Gujarat", "Maharashtra"]
growth = [342239, 94163, 308245, 196024, 307713]

# Create a pandas dataframe
data = pd.DataFrame({"State": State,
				"Growth": growth})

# Display Dataframe
data
```

```python
sns.barplot(x='State', y='Growth', data = data)
plt.show()
```

```python
# make barplot and sort bars
sns.barplot(x='State',
			y="Growth", data=data,
			order=data.sort_values('Growth').State)
```
