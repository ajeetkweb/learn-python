```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
iris  = pd.read_csv("../Iris.csv")
iris.head(5)
```

```python
iris['Species'].value_counts()
```

```python
sns.scatterplot(x='SepalLengthCm', y='PetalLengthCm', data = iris)
plt.show()
```

```python
sns.scatterplot(x = iris['SepalLengthCm'], y = iris['PetalLengthCm'])
plt.show()
```

```python
sns.scatterplot(x = iris['SepalLengthCm'], y = iris['PetalLengthCm'], data = iris, hue = 'Species')
plt.show()
```
