##### Linear Regression for House Price Prediction using Scikit-learn

#### 1. Import Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

##### Step 2: Load Dataset

```python
data = {
    'Area': [1000, 1500, 2000, 2500, 3000],
    'Price': [300000, 400000, 500000, 600000, 700000]
}
df = pd.DataFrame(data)
print(df)
```

##### step 3: Data Exploration

```python
sns.scatterplot(data=df, x='Area', y='Price')
plt.title('House Price vs Area')
plt.show()
```

#### Step 4: Prepare Data for Model

```python
# Features and target

x = df[['Area']]
y = df['Price']

# Splitting the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42) 
```

####  Step 5: Train the Linear Regression Model

```python
# instantiate the model
model = LinearRegression()

# Fit the model to the training data
model.fit(X_train, y_train)

# View the Model Parameters
print(f"Intercept: {model.intercept_}")
print(f"Coefficient: {model.coef_[0]}")


# Price = (Coefficient × Area) + Intercept
```

##### step 6: Make Predictions

```python
y_pred = model.predict(X_test)
print("Predicted Prices:", y_pred)
```

#### Step 7: Evaluate the Model

```python
mse = mean_squared_error(y_test, y_pred)
print("Actual Prices:", y_test.values)
print("Predicted Prices:", y_pred)
print(f"Mean Squared Error: {mse}")

```

#####  Step 8: Predict New Values

```python
new_area = np.array([[1800]])  # Area in square feet
predicted_price = model.predict(new_area)
print(f"Predicted price for 1800 sq.ft: ₹{predicted_price[0]:,.2f}")
```
