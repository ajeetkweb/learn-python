#### Python’s context managers provide a neat way to automatically set up and clean up resources.
#### __enter__() → sets up the context and returns the resource.
#### __exit__() -> handles cleanup (executed whether exceptions occur or not)

##### Context Managers ensure :
##### Ensure files are closed after reading/writing.
##### Ensure database connections are closed properly.

```python
#  Using Context Manager with Files

with open('sample.txt', 'w') as file:
    file.write("Hello, World!\n")
    file.write("This is a sample file.\n")
```

##### Creating a Custom Context Manager Class

```python
class ContextManager:

    def __init__(self):
        print('init method called')


    def __enter__(self):
        print('enter method called')
        return self
    

    def __exit__(self, exe_type, exe_value, exe_traceback):
        print('exit metho called')


with ContextManager() as manager:
    print('with statement block')
```

### file operation using context manager

```python
class FileManager:

    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
    
    def __enter__(self):
        self.file = open(self.filename, self.mode)
        return self.file
   
    def __exit__(self, exe_type, exe_value, exe_traceback):

        self.file.close()
        

with FileManager('sample.txt', 'r') as f:
    print(f.read())

print('File closed: ',f.closed)
```

### Database Connection Management with Context Manager

```python
from pymongo import MongoClient

class MongoDBConnectionManager:
    def __init__(self, hostname, port):
        self.hostname = hostname
        self.port = port
        self.connection = None

    def __enter__(self):
        self.connection = MongoClient(self.hostname, self.port)
        return self.connection

    def __exit__(self, exc_type, exc_value, exc_traceback):
        self.connection.close()

with MongoDBConnectionManager('localhost', 27017) as mongo:
    collection = mongo.SampleDb.test
    data = collection.find_one({'_id': 1})
    print(data.get('name'))
```
