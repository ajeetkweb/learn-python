####  Example 1: Basic Coroutine

```python
import asyncio

async def say_hello():
    print("Hello")
    await asyncio.sleep(1)
    print("World")

# asyncio.run(say_hello())
await say_hello()
```

##### Running Tasks Concurrently

```python
#Running Tasks Concurrently

import asyncio

async def task(name, delay):
    print(f"starting task {name}")
    await asyncio.sleep(delay)
    print(f"task {name} completed")


async def main():

    task1 = asyncio.create_task(task("A", 2))
    task2 = asyncio.create_task(task("B", 1))

    await task1
    await task2

await main()

# Task B finishes first even though it started after A 
```

##### gather() to Run Multiple Tasks

```python
# gather()

import asyncio

async def task(name, delay):
    await asyncio.sleep(delay)
    return f"Done with {name}"

async def main():
    results = await asyncio.gather(
        task("one", 2),
        task("two", 1),
        task("three", 3)
    )
    print(results)

# asyncio.run(main())
await main()
```

#####  Simulating Multiple Downloads

```python
import asyncio

async def download_file(file_name, delay):
    print(f"Starting download for {file_name}")
    await asyncio.sleep(delay)  # simulate I/O-bound delay
    print(f"Download completed for {file_name}")

async def main():

    # create multiple conccurent tasks
    task1 = asyncio.create_task(download_file('file1', 2))
    task2 = asyncio.create_task(download_file("file2", 1))
    task3 = asyncio.create_task(download_file("file3", 3))

    result = await asyncio.gather(task1, task2, task3)



# asyncio.run(main())
await main() # Simulating Multiple Downloads
```
