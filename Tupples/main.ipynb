##### A tuple is an immutable (unchangeable) and ordered collection in Python. It is defined using parentheses ()

```python
t = (10, 20, 30, 40)
print(len(t))  # Output: 4
```

```python
# max() - Get Maximum Value

t = (5, 9, 2, 7)
print(max(t))  # Output: 9
```

```python
# min() - Get Minimum Value

t = (5, 9, 2, 7)
print(min(t))  # Output: 2
```

```python
# sum() - Sum of Elements

t = (1, 2, 3, 4)
print(sum(t))  # Output: 10
```

```python
# sorted() - Sort Tuple Elements

t = (10, 3, 5, 8)
print(sorted(t))  # Output: [3, 5, 8, 10]
```

```python
# tuple() - Convert a List to Tuple

lst = [1, 2, 3]
print(tuple(lst))  # Output: (1, 2, 3)

s = "hello"
print(tuple(s))  # Output: ('h', 'e', 'l', 'l', 'o')
```

```python
# count() - Count Occurrences of an Element

t = (1, 2, 2, 3, 4, 2)
print(t.count(2))  # Output: 3
```
