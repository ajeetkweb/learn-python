#### An iterator in Python is an object that holds a sequence of values.
#### The iter() function returns an iterator from an iterable.
####  The next() function retrieves the next item from the iterator.

#### Basic Iterator Usage

```python
nums = [1, 2, 3]
it = iter(nums)  # returns iterator

print(next(it))  # 1
print(next(it))  # 2
print(next(it))  # 3
# next(it) -> raises StopIteration
```

##### Creating custom iterator

```python
class EvenNumbers:

    def __init__(self, limit = 10):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):

        if self.current < self.limit:
            result = self.current
            self.current += 2
            return result
        else:
            raise StopIteration    


# Create an instance of EvenNumbers
even = EvenNumbers()
# Iterate through the even numbers
for num in even:
    print(num)  # Output: 0, 2, 4, 6, 8
```

#### StopIteration Exception

```python
nums = [1, 2, 3, 4, 5]
it = iter(nums)

while True:
    try:
        print(next(it))
    except StopIteration:
        print("Reached the end of the iterator.")
        break
```
