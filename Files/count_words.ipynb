# Find the count of given word in text file 

```python
counter = 0
text = 'hello'
with open("hello.txt", 'r') as file:
    content = file.read().lower()

    for word in content.split():
        if word == text:
            counter += 1

print(counter)
```

```python
counter = 0
text = 'hello'

with open("hello.txt", 'r') as file:
    for line in file:
        for word in line.split():
            if word.lower() == text:
                counter += 1
print(counter)
```
