# Pie chart
## A Pie Chart is a circular statistical plot that can display only one series of data

### Syntax: matplotlib.pyplot.pie(data, explode=None, labels=None, colors=None, autopct=None, shadow=False)

```python
# import modules
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Creating dataset
cars = ['AUDI', 'BMW', 'FORD',
        'TESLA', 'JAGUAR', 'MERCEDES']
 
data = [23, 17, 35, 29, 12, 41]
```

```python
# Creating plot
fig = plt.figure(figsize =(6, 4))
plt.pie(data, labels = cars)
 
# show plot
plt.show()
```

## Customizing Pie Chart

```python
plt.pie(data, labels = cars, autopct = '%1.1f%%', shadow = True, startangle=90)
plt.show()
```
