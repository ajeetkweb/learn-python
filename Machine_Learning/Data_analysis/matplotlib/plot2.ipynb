```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0,5,11)
y = x**2
x
```

```python
y
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,0.8,0.8])
ax.plot(x,y, color ='orange', lw=2, ls = '-', marker = 'o', markersize = 18, markerfacecolor = "blue")
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,0.8,0.8])
ax.scatter(x,y, color = 'r')
ax.set_title('Scatter plot')
```
