## sns.boxplot()

### seaborn.boxplot(data=None, *, x=None, y=None, hue=None, color=None, palette=None, saturation=0.75, width=0.8, dodge=True)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../Titanic-Dataset.xls")
df.head(2)
```

```python
sns.boxplot(x=df["Age"])
```

```python
sns.boxplot(data=df, x="Age", y="Pclass", hue = 'Sex')
```

```python
sns.boxplot(data=df, x="Age", y="Pclass", hue="Survived")
```
