##### Find All Pairs With a Given Sum

```python
def find_pairs(nums, target):
    pairs = []

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.append((nums[i], nums[j]))
                    
    return pairs


nums = [1, 2, 3, 4, 5]
target = 5
print(find_pairs(nums, target)) 
```

### Two Sum Problem

```python
def two_sum(nums, target):
    seen = {}
    pairs = []

    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            pairs.append((complement, num))

        seen[num] = i
    
    return pairs

nums = [1, 2, 3, 4, 5]
target = 5
result = two_sum(nums, target)
print(result)
```

##### Flatten a Nested List


```python
def flatten(nested_list):
    flat_list = []

    for item in nested_list:
        if isinstance(item, list):
            flat_list.extend(flatten(item))
        else:
            flat_list.append(item)
    
    return flat_list    



print(flatten([1, [2, [3, 4], 5], 6]))
```

### Merge Two sorted List

```python
def merge_sorted_list(a, b):

    i = 0
    j = 0
    c = []

    while i < len(a) and j < len(b):

        if a[i] < b[j]:
            c.append(a[i])
            i += 1
        
        else:
            c.append(b[j])          
            j += 1
    
    c.extend(a[i:])
    c.extend(b[j:])
    return c


a = [1, 3, 5]
b = [2, 4, 6]
merged_list = merge_sorted_list(a, b)
print(merged_list)
```

### Find Second Largest Number

```python
def sort_list(lst):
    return sorted(lst)

def second_largest_number(lst):

    sorted_list = sort_list(lst)
    num = sorted_list[len(sorted_list) -2]
    return num


my_list = [10, 20, 4, 45, 99]
num = second_largest_number(my_list)
print(num)
```
