### 1. Fibonacci Using Memoization (Top-Down DP)
results are stored in a dictionary.

   Time Complexity: O(n) <br>
✔ Space Complexity: O(n)

```python
def fibonacci(n, memo = {}):
    
    if n in memo:
        return memo[n]
    
    if n <= 1:
        return n
    
    memo[n] = fibonacci(n-1, memo) + fibonacci(n-2, memo)
    return memo[n]

print(fibonacci(10))
```
