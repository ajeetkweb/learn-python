### Reverse a string

```python
def reversed(name):

    reversed_name = ''
    for char in name:
        reversed_name = char + reversed_name
    return reversed_name


name = 'Ajeetkumar'
result = reversed(name)
print(result)
```

### Reverse words in a sentence

```python
def reverse_words(sent):

    words = sent.split()
    reversed_sent = ''
    for word in words:
        reversed_sent = word + ' ' + reversed_sent
    return reversed_sent

sent = 'Python is a great language'
result = reverse_words(sent)
print(result)
```

### Check if a string is a palindrome

```python
def is_palindrome(s):
    left = 0
    right = len(s) - 1
    while left < right:
        if s[left] != s[right]:
            return False
        left += 1
        right -= 1
        
    return True

print(is_palindrome("radar"))  # Output: True
```

####  Check if Two Strings are Anagrams
##### Logic: : Count each character manually in both strings.

```python
def is_anagram(s1, s2):

    if len(s1) != len(s2):
        return False
    
    count1 = {}
    count2 = {}

    for char in s1:
        count1[char] = count1.get(char, 0) + 1
        
    for char in s2:
        count2[char] = count2.get(char, 0) + 1
    
    return count1 == count2




print(is_anagram("listen", "silent"))  # Output: True
```

##### Group Anagrams from Given List

```python
def group_anagrams(words):

    anagrams = {}

    for word in words:
        sorted_word = ''.join(sorted(word))


        if sorted_word in anagrams:
            anagrams[sorted_word].append(word)
        else:
            anagrams[sorted_word] = [word]
    
    res = list(anagrams.values())
    return res

w = ["eat", "tea", "tan", "ate", "nat", "bat"] 
res = group_anagrams(w)
print(res)
```

### Count vowels and consonants in a string

```python
def count_vowels_and_consonants(name):

    vowels = 'aeiouAEIOU'
    vowel_count = 0
    consonant_count = 0

    for char in name:
        if char in vowels:
            vowel_count += 1
        elif char.isalpha():
            consonant_count += 1
   
    
    print(f"Vowels: {vowel_count}, Consonants: {consonant_count}")


name = 'Ajeet5kumar6'
count_vowels_and_consonants(name)
```

#### Remove duplicate characters from a string

```python
def remove_duplicates(name):
    new_name  = ''
    for char in name:
        if char not in new_name:
            new_name += char
    return new_name



name = 'ajeetkumar'
# output : ajetkumr
result = remove_duplicates(name)
print(result)
```

### Find the frequency of characters in a string

```python

def find_frequency(name):
    frequency = {}
    for char in name:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1
    return frequency


res = find_frequency('ajeetkumar')
print(res)
```

#### Find the first non-repeating character


```python
def first_non_repeating_char(name):

    frequency = {}

    for char in name:
        frequency[char] = frequency.get(char, 0) + 1

        for char in name:
            if frequency[char] == 1:
                return char
    return None


print(first_non_repeating_char("programming"))
```
