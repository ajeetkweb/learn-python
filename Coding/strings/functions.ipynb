#### count()
#### Returns the number of occurrences of a substring in the given string.

```python
## string.count()

message = 'python is popular programming language.python is easy to learn'
print('No of occurrance of "python" in the message is:', message.count('python'))
```

##### find() : 
##### Returns the index of first occurrence of the substring (if found). If not found, it returns -1.

```python
# string.find()

message = 'Python is a fun programming language'

# check the index of 'fun'
print(message.find('fun'))

# Output: 12
```

#####  format()

```python
# default arguments
print("Hello {}, your balance is {}.".format("Adam", 230.2346))

# positional arguments
print("Hello {0}, your balance is {1}.".format("Adam", 230.2346))

# keyword arguments
print("Hello {name}, your balance is {blc}.".format(name="Adam", blc=230.2346))

# mixed arguments
print("Hello {0}, your balance is {blc}.".format("Adam", blc=230.2346))
```

##### isalpha() : Return True if all characters in the string are alphabets.

```python
# string.isalpha()

name = "Monica"
print(name.isalpha())

# contains whitespace
name = "Monica Geller"
print(name.isalpha())

# contains number
name = "Mo3nicaGell22er"
print(name.isalpha())
```

```python
# isdigit() :  Returns True if all characters in a string are digits. 
# If not, it returns False.

str1 = '342'
print(str1.isdigit())

str2 = 'python'
print(str2.isdigit())

# Output: True
#         False

```

```python
# string.islower()

s = 'this is good'
print(s.islower())

s = 'this is Not good'
print(s.islower())


```

```python
# isnumeric():  method checks if all the characters in the string are numeric.

pin = "523"

# checks if every character of pin is numeric 
print(pin.isnumeric())

```

####  join():  Returns a string by joining all the elements of an iterable (list, string, tuple), separated by the given separator.

```python
#  join()

text = ['Python', 'is', 'a', 'fun', 'programming', 'language']

print(' '.join(text))

```

```python
# string.lower()

message = 'PYTHON IS FUN'

# convert message to lowercase
print(message.lower())
```

```python
# string.upper()

message = 'python is fun'

# convert message to uppercase
print(message.upper())
```

```python
# string.replace()

text = 'bat ball'

# replace 'ba' with 'ro'
replaced_text = text.replace('ba', 'ro')
print(replaced_text)
```

#### split() : method breaks down a string into a list of substrings using a chosen separator.

```python
text = 'Python is fun'

# split the text from space
print(text.split())

# Output: ['Python', 'is', 'fun']
```
