Merge Two Sorted Lists

```python
def merge_sorted_list(a, b):

    i = 0
    j = 0
    c = []

    while i < len(a) and j < len(b):

        if a[i] < b[j]:
            c.append(a[i])
            i += 1
        
        else:

            c.append(b[j])
            j += 1
    
    c.extend(a[i:])
    c.extend(b[j:])
    return c


a = [1, 3, 5]
b = [2, 4, 6]
merged_list = merge_sorted_list(a, b)
print(merged_list)
```

Find Second Largest Element

```python
def sort_list(lst):
    return sorted(lst)

def second_largest_number(lst):

    sorted_list = sort_list(lst)
    num = sorted_list[len(sorted_list) -2]
    return num

    

my_list = [10, 20, 4, 45, 99]

num = second_largest_number(my_list)
print(num)
```
