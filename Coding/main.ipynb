Merge Two Sorted Lists

```python
def merge_sorted_list(a, b):

    i = 0
    j = 0
    c = []

    while i < len(a) and j < len(b):

        if a[i] < b[j]:
            c.append(a[i])
            i += 1
        
        else:

            c.append(b[j])
            j += 1
    
    c.extend(a[i:])
    c.extend(b[j:])
    return c


a = [1, 3, 5]
b = [2, 4, 6]
merged_list = merge_sorted_list(a, b)
print(merged_list)
```

Find Second Largest Element

```python
def sort_list(lst):
    return sorted(lst)

def second_largest_number(lst):

    sorted_list = sort_list(lst)
    num = sorted_list[len(sorted_list) -2]
    return num

    

my_list = [10, 20, 4, 45, 99]

num = second_largest_number(my_list)
print(num)
```

#### Two Sum problems

```python
def two_sum(nums, target):
    lookup = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in lookup:
            return [lookup[diff], i]
        lookup[num] = i

nums = [5, 7, 9, 4]
target = 16
result = two_sum(nums, target)
print(result)
```

#### Valid Anagram

##### two strings are considered anagrams if they have the same characters with the same frequency, but the order of the characters is different.

#####  “listen” and “silent” are anagrams 

```python

def is_anagram(s, t):

    if len(s) != len(t):
        return False
    
    count = {}

    for char in s:
        count[char] = count.get(char, 0) + 1
    
    for char in t:
        if char not in count or count[char] == 0:
            return False
        
        count[char] -= 1
    
    return True




s = "silent"
t = "listen"

result = is_anagram(s, t)
print(result)
```

##### Group Anagrams from Given List

```python

def group_anagrams(words):

    anagrams = {}

    for word in words:
        sorted_word = ''.join(sorted(word))

        if sorted_word in anagrams:
            anagrams[sorted_word].append(word)
        else:
            anagrams[sorted_word] = [word]
    
    res = list(anagrams.values())
    return res

w = ["eat", "tea", "tan", "ate", "nat", "bat"] 
res = group_anagrams(w)
print(res)
```

#### Longest Substring Without Repeating Characters

```python
def length_of_longest_substring(s):
    seen = {}
    left = max_len = 0
    for right, char in enumerate(s):
        if char in seen and seen[char] >= left:
            left = seen[char] + 1
        seen[char] = right
        max_len = max(max_len, right - left + 1)
    return max_len
```

```python
print(length_of_longest_substring("abcdeabcbb"))  # Output: 3
```

##### Write a Python code to implement a binary search algorithm


```python
def binary_search(arr, target):
    low = 0
    high = len(arr) - 1

    while low <= high:
        mid = (low + high) // 2  # Find the middle index
        guess = arr[mid]

        if guess == target:
            return mid  # Target found at index mid
        elif guess < target:
            low = mid + 1  # Search in the right half
        else:
            high = mid - 1  # Search in the left half

    return -1  # Target not found

# Example usage
arr = [1, 3, 5, 7, 9, 11, 13, 15]
target = int(input("Enter the number to search for: "))

result = binary_search(arr, target)

if result != -1:
    print(f"Element {target} found at index {result}.")
else:
    print(f"Element {target} not found in the array.")
```

##### bubble sort 

```python
def bubble_sort(elements):
    n = len(elements)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if elements[j] > elements[j + 1]:
                elements[j], elements[j + 1] = elements[j + 1], elements[j]

# Test the function
nums = [5, 2, 8, 1, 9]
bubble_sort(nums)
print(nums)
```
