Filter palindrome words

```python
words = ["madam", "python", "level", "world"]
result = list(filter(lambda x: x == x[::-1], words))

print(result)
```

map() Function <br>

map() applies a function to every element in an iterable (like a list). <br>

map(function, iterable)

```python
# Square numbers using lambda + map

numbers = [1, 2, 3, 4, 5]

result = list(map(lambda x: x**2, numbers))

print(result)
```

filter(function, iterable) <br>

filter() selects elements that satisfy a condition.



```python
numbers = [1,2,3,4,5,6,7,8]

evens = list(filter(lambda x: x % 2 == 0, numbers))

print(evens)
```

```python
# Square of even numbers

numbers = [1,2,3,4,5,6]

result = list(map(lambda x: x**2,
           filter(lambda x: x%2==0, numbers)))

print(result)
```
