##### A set in Python is an unordered collection of unique elements, defined using curly braces {} or the set() function

```python
my_set = {1, 2, 3, 4}
print(my_set)  # Output: {1, 2, 3, 4}
```

```python
# Using set() Constructor

my_set = set([1, 2, 3, 3, 4])  # Duplicate '3' is removed
print(my_set)  # Output: {1, 2, 3, 4}
```

```python
# add() - Add an Element

s = {1, 2, 3}
s.add(4)
print(s)  # Output: {1, 2, 3, 4}
```

```python
# remove() - Remove an Element

s = {1, 2, 3}
s.remove(2)
print(s)  # Output: {1, 3}
```

#### Mathematical Set Operations


```python
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))  # Output: {1, 2, 3, 4, 5}
print(a | b)       # Output: {1, 2, 3, 4, 5}
```

```python
# intersection() or & - Common Elements

a = {1, 2, 3}
b = {3, 4, 5}
print(a.intersection(b))  # Output: {3}
print(a & b)              # Output: {3}
```

```python
## difference() or - - Elements in A not in B
a = {1, 2, 3}
b = {3, 4, 5}
print(a.difference(b))  # Output: {1, 2}
print(a - b)            # Output: {1, 2}
```
