##### Concatenating (pd.concat)

##### Adding new rows or columns to a dataset.

##### DataFrames have same columns (vertical stack) or same index (horizontal stack).

```python
import pandas as pd

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})

# Vertically concatenate
result_v = pd.concat([df1, df2], axis=0, ignore_index=True)

print("Result of vertical concatenation:")
print(result_v)

# Horizontally concatenate
result_h = pd.concat([df1, df2], axis=1)
print("Result of horizontal concatenation:")
print(result_h)
```

#### GroupBy and Aggregation

##### Aggregation means applying a mathematical function to summarize data. It can be used to get a summary of columns in our dataset like getting sum, minimum, maximum etc.

##### sum()         : Compute sum of column values in each column
##### min()          : Compute min of column values
##### max()         : Compute max of column values
##### mean()       : Compute mean of column

```python
df = pd.DataFrame([[9, 4, 8, 9],
                   [8, 10, 7, 6],
                   [7, 6, 8, 5]],
                  columns=['Maths',  'English', 
                           'Science', 'History'])

print(df)
```

```python
#  sum() 

print(df.sum())  # sum of each column
```

```python
# describe()

df.describe() # summary statistics
```

```python
# agg()

df.agg(['sum', 'min', 'max'])   # apply multiple functions
```

#### Grouping in Pandas :   groupby('column')

##### Grouping in Pandas means organizing your data into groups based on some columns.

```python
import pandas as pd

# small bakery order dataset
data = {
    'Item': ['Cake', 'Cake', 'Bread', 'Pastry', 'Cake'],
    'Flavor': ['Chocolate', 'Vanilla', 'Whole Wheat', 'Strawberry', 'Chocolate'],
    'Price': [250, 220, 80, 120, 250]
}

df = pd.DataFrame(data)
print(df)
```

```python
grouped = df.groupby('Item')    # Group by 'Item' column
print(grouped)      #  it just creates a grouped object
```

```python
#  output shows the total earnings from each item

print(df.groupby('Item')['Price'].sum()) # the total price of each item sold:
```

```python
# Grouping by Multiple Columns

print(df.groupby(['Item', 'Flavor'])['Price'].sum())
```

##### concat()

```python
df1 = pd.DataFrame(
    {
        "A": ["A0", "A1", "A2", "A3"],
        "B": ["B0", "B1", "B2", "B3"],
        "C": ["C0", "C1", "C2", "C3"],
        "D": ["D0", "D1", "D2", "D3"],
    },
    index=[0, 1, 2, 3],
)
df2 = pd.DataFrame(
    {
        "A": ["A4", "A5", "A6", "A7"],
        "B": ["B4", "B5", "B6", "B7"],
        "C": ["C4", "C5", "C6", "C7"],
        "D": ["D4", "D5", "D6", "D7"],
    },
    index=[4, 5, 6, 7],
)

frames = [df1, df2]

result = pd.concat(frames)
print(result)
```
