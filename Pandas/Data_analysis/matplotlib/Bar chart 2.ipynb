### runs scored by former Indian Captain MS Dhoni across all the formats, and then we will compare the percentage of runs scored across all formats with respect to the total runs scored by him

```python
# import modules
import pandas as pd
import matplotlib.pyplot as plt
```

```python
dt = {
    'Format':  ['Test', 'ODI', 'T20I', 'IPL'],
    'Matches': [90, 350, 98, 204],
    'Runs'   : [4876, 10773, 1617, 4632]
 }

data = pd.DataFrame(dt)
data
```

## No of matches played by dhoni in different formats

```python
fig = plt.figure(figsize = (6,6))

plt.bar(data['Format'], data['Matches'])
plt.title("No of Matches played by Dhoni in defferent Formats")
plt.xlabel("Formats")
plt.ylabel('Matches played')
plt.show()
```

## Runs made by Dhoni in diffrent formats

```python
fig = plt.figure(figsize = (5,3))

colors_list = ['Red','Orange', 'Blue', 'Purple']
plt.bar(data['Format'], data['Runs'], color = colors_list)
plt.title("Runs made by Dhoni in diffrent formats")
plt.xlabel("Formats")
plt.ylabel('Runs')
plt.show()
```

## Total matches played

```python
data['Matches'].sum()
```

## Total runs scored

```python
total_runs = data['Runs'].sum()
total_runs
```

```python
percentage = []

for i in range(data.shape[0]):
    prct = (data.Runs[i] / total_runs) * 100
    percentage.append(round(prct, 2))

percentage
```

```python
data['Percentage'] = percentage
data
```

```python
plt.figure(figsize=(8,8))
colors_list = ['Red','Orange', 'Blue', 'Purple']
graph = plt.bar(data.Format,data.Runs, color = colors_list)
plt.title('Percentage of runs scored by MS Dhoni across all formats')


i = 0
for p in graph:
    width = p.get_width()
    height = p.get_height()
    x, y = p.get_xy()
    plt.text(x+width/2,
             y+height*1.01,
             str(data.Percentage[i])+'%',
             ha='center',
             weight='bold')
    i+=1
   

plt.show()

```
