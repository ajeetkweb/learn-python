### conditional filtering

```python
import pandas as pd
```

```python
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'Los Angeles', 'Chicago'],
        'Salary': [70000, 80000, 90000],
        'ABC' : [1, 2, 3]
        }

df = pd.DataFrame(data)
print(df)
```

```python
df2 = df[df['Salary'] > 75000]
print(df2)
```

```python
df3 = df[df['City'] == 'New York']
print(df3)
```

```python
df4 = df[(df['Salary'] > 75000) & (df['City'] == 'Los Angeles')]
print(df4)
```

```python
df5 = df[(df['Salary'] > 75000) | (df['City'] == 'Chicago')]
print(df5)
```

```python
df7 = df[df['City'].isin(['New York', 'Chicago'])]
print(df7)
```

##### Using apply() with a Lambda Function

```python
# apply()

data = {'Name': ['John', 'Sophia', 'Daniel', 'Emma'],
        'Experience': [5, 8, 3, 10]}
df = pd.DataFrame(data)
print("==== Original Dataset ======")
print(df)

# Apply if condition using lambda function

df['Category'] = df['Experience'].apply(lambda x: 'Senior' if x > 5 else 'Junior')

print("==== Dataset with 'Senior'and 'Junior' Category =====")
print(df)
```

#### Using np.where()

```python
# Sample DataFrame with student scores
data = {'Student': ['Alice', 'Bob', 'Charlie', 'David'],
        'Score': [75, 40, 85, 60]}

df = pd.DataFrame(data)
print(df)

df['Result'] = df['Score'].apply(lambda x: 'Pass' if x >= 50 else 'Fail')
print("==== Dataset with 'Pass' and 'Fail' =====")
print(df)

# using np.where()
import numpy as np

data['Result'] = np.where(df['Score'] >= 50, 'Pass', 'Fail')
print("==== Dataset with 'Pass' and 'Fail' using np.where() =====")
print(df)
```

#### Using loc[] for Applying 'If Condition'

```python
data = {'Product': ['Laptop', 'Phone', 'Tablet', 'Smartwatch'],
        'Price': [1200, 600, 300, 200]}


df = pd.DataFrame(data)

df.loc[df['Price'] > 500, 'Category'] = 'Expensive'
df.loc[df['Price'] <= 500, 'Category'] = 'Affordable'

print(df)
```

##### Using query() for Filtering Data Based on Multiple Conditions

##### query() method is a great tool for filtering rows based on multiple conditions.

```python
# query()

data = {'Employee': ['John', 'Sophia', 'Daniel', 'Emma'],
        'Experience': [6, 2, 7, 4],
        'Salary': [55000, 40000, 66000, 35000]
        }

df = pd.DataFrame(data)

filtered_df = df.query('Experience >= 5 and Salary > 50000')

print(filtered_df)
```
