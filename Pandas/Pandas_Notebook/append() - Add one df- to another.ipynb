```python

# Syntax of append()

# DataFrame.append(other, ignore_index=False, verify_integrity=False, sort=False)

# used to append one DataFrame row(s) and column(s) with another,
```

```python

import pandas as pd

df = pd.DataFrame({'Courses': ["Spark","PySpark","Python","pandas"],
                    'Fee' : [20000,25000,22000,24000]})

df1 = pd.DataFrame({'Courses': ["Pandas","Hadoop","Hyperion","Java"],
                    'Fee': [25000,25200,24500,24900],
                    'Duration': ['30days','35days','40days','45days']})

# Using append() method
df2 = df.append(df1)
print(df2)
```

```python
# Reindex the DataFrame while Append
# ignore_index = True

df2 = df.append(df1, ignore_index = True)
df2
```

```python
# Create third DataFrame  
df2 = pd.DataFrame({'Courses':['PHP','GO'],
                    'Duration':['30day','40days'],
                    'Fee':[10000,23000]})
df3 = df.append([df1, df2], ignore_index = True)
df3
```
