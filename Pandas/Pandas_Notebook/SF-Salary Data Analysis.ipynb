## What is the name of highest paid person (including benefits)?

```python
sal[sal['TotalPayBenefits'] == sal['TotalPayBenefits'].max()]
```

```python
sal[sal['TotalPayBenefits'] == sal['TotalPayBenefits'].min()]
```

## What was the average (mean) TotalPayBenefits of all employees per year? (2011-2014) ?

```python
sal.groupby('Year').mean()['TotalPayBenefits']
```

## How many people have the word Chief in their job title?

```python
def chief(title):
    if "chief" in title.lower():
        return True
    else:
        return False
```

```python
sum(sal['JobTitle'].apply(lambda x: chief(x)))
```

```python
sal[sal['JobTitle'] == 'Special Nurse']
```

## How many Special Nurse are there in job title?

```python
def specialNurse(job_title):
    if "Special Nurse" in job_title:
        return True
    else:
        return False


sum(sal['JobTitle'].apply(lambda x: specialNurse(x)))
```
