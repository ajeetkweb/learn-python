```python
import pandas as pd

# create DataFrmae from a List

technologies = [['PySpark', 500000, '60 days'],
                ['Pandas', 40000, '30 days'],
                ['Python', 60000, '90 days']
               ]
df = pd.DataFrame(technologies)
df
```

```python
# create dataFrame with columns

column_names = ['Courses', 'Fee', 'Duration']
df1 = pd.DataFrame(data = technologies, columns = column_names)
df1
```

```python
# create Empty DataFrame with columns

df = pd.DataFrame(columns = ['Skills','Exp', 'Rating'])
df
df.info()
```

```python
# create DataFrame from a Dict (Dictionary)

techs = {
        'Courses': ['Pyspark', 'Pandas', 'Python', 'MySQL'],
        'Fee'    : [5000, 3000, 8000, 6000],
        'Duration': ['30 days', '60 days', '90 days', '120 days']
    }

df = pd.DataFrame(techs)
df
```

```python
df.info()
```

```python
# change data Types of dataframe columns
# astype()

types = {'Courses': str, 'Fee': int, 'Duration': str}
df = df.astype(types)
df.info()
```
