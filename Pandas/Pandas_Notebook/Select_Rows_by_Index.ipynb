```python
import pandas as pd

techs = {
        'Courses': ['Pyspark', 'Pandas', 'Python', 'MySQL'],
        'Fee'    : [5000, 3000, 8000, 6000],
        'Duration': ['30 days', '60 days', '90 days', '120 days']
    }

df = pd.DataFrame(techs)
df
```

```python
# Select Rows by Index 

# iloc()
# loc()

df2 = df.iloc[2]
df2
```

```python
# Select Rows by Index List

df2 = df.iloc[[1,0,2]]
df2
```

```python
# Select Last Row

df2 = df.iloc[-1]
df2
```

```python
# Select First 2 Rows

df2 = df.iloc[:2]
df2
```

```python
# Select First 2 Rows and first two columns

df3 = df.iloc[:2, [0,1]]
df3
```

```python


import numpy as np
technologies = {
    'Courses':["Spark","PySpark","Hadoop","Python","pandas","Oracle","Java"],
    'Fee' :[20000,25000,26000,22000,24000,21000,22000],
    'Duration':['30days','40days','35days','40days',np.nan,None,'55days'],
    'Discount':[1000,2300,1500,1200,2500,2100,2000]
               }
index_labels=['r1','r2','r3','r4','r5','r6','r7']
df = pd.DataFrame(technologies,index=index_labels)
print(df)
```

```python
# loc[]  - Select Rows by Index Labels


df2 = df.loc['r7']
df2
```
