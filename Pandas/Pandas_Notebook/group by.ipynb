```python
import numpy as np
import pandas as pd
```

```python
data = { 'Company' : ['Facebook','Google','Microsoft','Facebook','Google','Google','Microsoft','Facebook','Microsoft'],
         'Person'  : ['Sam', 'Ram', 'Amy', 'Carl', 'Sarah', 'Mohan', 'Mahesh', 'Uma', 'Ajit'],
         'Sales'   : [200, 130, 160, 300, 500, 450, 600, 100, 50]   
}

df = pd.DataFrame(data)
```

```python
df
```

## groupby

```python
groupByComp = df.groupby('Company')
```

```python
# perform Aggregate functions

groupByComp.sum()
```

```python
groupByComp.mean()
```

```python
groupByComp.std()
```
