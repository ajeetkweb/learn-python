```python
# DataFrame dropna() Usage 

import pandas as pd
import numpy as np

technologies= {
    'Courses':["Spark","PySpark","Hadoop","Python",np.nan],
    'Fee' :[22000,25000,23000,24000,np.nan],
    'Duration':['30days','50days','30days', None,np.nan],
    'Discount':[1000,2300,1000,1200,np.nan]
          }
df = pd.DataFrame(technologies)
print(df)
```

```python
df.isna().sum()
```

```python
# Drop rows that has all NaN values

df = df.dropna(how = 'all')
df
```

```python
df.isna().sum()
```

```python
df.assign(TutorAssigned = np.nan)
```

```python
# Drop columns that has all NaN values

df = df.dropna(how='all', axis = 1)
df
```

```python
# Default drop rows that contains nan values

df2=df.dropna()
df2
```

```python
# Drop all columns with NaN values
df2=df.dropna(axis=1)
df2
```

```python
 # Drop rows that has NaN values on selected columns
    
df2=df.dropna(subset=['Courses','Duration'])
df2
```
