```python
#  DataFrame.insert() -  Add Column to DataFrame

# assign()  - returns a new DataFrame after adding a column


# Create DataFrame
import pandas as pd
import numpy as np

technologies= {
    'Courses':["Spark","PySpark","Hadoop","Python","Pandas"],
    'Fee' :[22000,25000,23000,24000,26000],
    'Discount':[1000,2300,1000,1200,2500]
          }

df = pd.DataFrame(technologies)
print(df)
```

```python

# Let’s add a column ‘TutorsAssigned” to the DataFrame using assign()

# Add new column to the DataFrame
tutors = ['William', 'Henry', 'Michael', 'John', 'Messi']
df2 = df.assign(TutorAssigned = tutors)
df2
```

```python
 # Add Multiple Columns to the DataFrame

MNCCompanies = ['Google', 'TCS', 'Tech Mahindra', 'Wipro', 'HCL']

df2 = df.assign(TutorAssigned = tutors, MNCComp = MNCCompanies, Address = np.nan)
df2
```

```python
# Adding a Column From Existing
# derives Discount_Percent column from Fee and Discount

df = pd.DataFrame(technologies)
df2 = df.assign(Discount_percent = lambda x: (x.Discount * 100) / x.Fee)
df2
```

```python
# Append Column to Existing Pandas DataFrame

```

```python
df = pd.DataFrame(technologies)
df['MNCComp'] = MNCCompanies
df
```

```python
# Add Column to Specific Position of DataFrame
# DataFrame.insert() method is used to add DataFrame at any position of the existing DataFrame.


# Add Tutors column at the beginning of the DataFrame.

df = pd.DataFrame(technologies)
print(df)
print('')

df.insert(0, 'Tutor', tutors)

print('Added Tutors column at the beginning')
df
```
