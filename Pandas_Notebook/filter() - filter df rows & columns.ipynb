```python
#  filter() - filters the DataFame rows and columns.

# Returned DataFrame contains only rows and columns that are specified with the function

# doesn’t update the existing DataFrame instead it always returns a new one.


# Syntax of DataFrame.filter()

# DataFrame.filter(items=None, like=None, regex=None, axis=None)
```

```python
# Create DataFrame
import pandas as pd
technologies= {
    'Courses':["Spark","PySpark","Spark","Java","PySpark","PHP"],
    'Fee' :[22000,25000,23000,24000,26000,27000],
    'Duration':['30days','50days','30days','60days','35days','30days']
          }
df = pd.DataFrame(technologies)
df
```

## filter() Columns by Labels

```python

# Filter columns
df2=df.filter(items=['Courses','Fee'])
print(df2)
```
