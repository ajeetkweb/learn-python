```python
# drop() method is used to remove the columns from the DataFrame
# returns a new DataFrame after dropping the columns specified with the drop method.
# to remove columns on the existing DataFrame object use inplace=True param

```

```python

# Create DataFrame
import pandas as pd
technologies = ({
    'Courses':["Spark","PySpark","Hadoop","Python","pandas","Oracle","Java"],
    'Fee' :[20000,25000,26000,22000,24000,21000,22000],
    'Duration':['30day', '40days' ,'35days', '40days', '60days', '50days', '55days']
              })
df = pd.DataFrame(technologies)
df
```

```python
# drop() - Drop Columns

df2 = df.drop(['Fee'], axis = 1)
df2
```

```python
df2 = df.drop(['Fee', 'Duration'], axis = 1)
df2
```

```python
#  Drop Column by Index

df2 = df.drop(df.columns[[1]], axis = 1)
df2
```

```python
df2 = df.drop(df.columns[[0,1]], axis = 1)
df2
```

```python
df.drop(['Fee', 'Duration'], axis = 1, inplace = True)
df
```
