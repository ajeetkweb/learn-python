```python
# Create DataFrame
import pandas as pd
import numpy as np
df = pd.DataFrame(({
     'Courses':["Spark",'Java',"Scala",'Python'],
     'Fee' :[20000,np.nan,26000,24000],
     'Duration':['30days','40days','NA','40days'],
     'Discount':[1000,np.nan,2500,None]
               }))
print(df)
```

```python
# pandas fillna() NaN with None Value

df2=df.fillna('None')
print(df2)
```

```python
# Fillna on one column

df2['Discount'] = df['Discount'].fillna('0')
print(df2)
```

```python
# fillna on Multiple Columns

df2[['Dsicount','Fee']] = df[['Discount', 'Fee']].fillna('0')
df2
```
