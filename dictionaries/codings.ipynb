### Python program to find the sum of all items in a dictionary

```python
# Using sum()

d = {'a': 100, 'b': 200, 'c': 300}
res = sum(d.values())
print(res)
```

```python
# Using list comprehension

d = {'a': 100, 'b': 200, 'c': 300}

res = sum([d[key] for key in d])
print(res)
```

```python
# Using loop

d = {'a': 100, 'b': 200, 'c': 300}
res = 0

for value in d.values():
    res += value
print(res)
```

#### Sort Python Dictionary by Key 

```python
d = {' Bami': 10, 'Ajnish': 9, 'sanjeev': 15}

myKeys = list(d.keys())
myKeys.sort()

# Sorted Dictionary
sd = {i: d[i] for i in myKeys}
print(sd)
```

#### Sort Python Dictionary by Value

```python
my_dict = {'apple': 10, 'banana': 5, 'orange': 8}

# Sort by value
sorted_dict = dict(sorted(my_dict.items(), key=lambda item: item[1]))

print(sorted_dict)
```

#### Remove all duplicates words from a given sentence

```python
# Using set with join()

s1 = "Geeks for Geeks"
s2 = s1.split()  # Split the sentence into words

# Convert the list to a set and back to a list to remove duplicates
s3 = list(set(s2))

# Join the list back into a sentence
s4 = ' '.join(s3)
print(s4)
```

```python
# Using Simple Loop

s1 = "Geeks for Geeks"
s2 = s1.split()
res = []  

for word in s2:
    if word not in res:
        res.append(word)


s3 = ' '.join(res)
print(s3)
```
